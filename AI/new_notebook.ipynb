from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()
client = OpenAI()

text = "그린에너지 원가율이 계속 오르는 이유와 하반기 개선 계획을 정리해 주세요."

resp = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
            "role": "system",
            "content": "다음 문장의 유형을 하나만 판단하라. 반드시 아래 네 가지 중 한 줄만 출력하고 다른 말은 하지 마라.\n경영진 질의\n거래처 클레임\nRFQ 문의\n은행 RM",
        },
        {"role": "user", "content": text},
    ],
    temperature=0,
)

print(resp.choices[0].message.content.strip())
